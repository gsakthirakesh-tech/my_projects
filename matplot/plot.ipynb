```python
#plotting
import matplotlib.pyplot as plt


```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
```

```python
plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# scatter plot
x = [5, 7, 8, 10, 12]
y = [10, 14, 15, 18, 20]
```

```python
plt.scatter(x,y,color='red',marker='o')
plt.title("scattering points")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
#vertical bar plot

x = ['Apple', 'Banana', 'Cherry']
y = [30, 45, 10]
```

```python
plt.bar(x,y,color="orange")
plt.title("vertical bar graph")
plt.xlabel("Fruits")
plt.ylabel("price")
plt.show()
```

```python
#horizontal bar plot

x = ['Python', 'Java', 'C++']
y = [90, 75, 60]
```

```python
plt.barh(x,y,color="red")
plt.title("horizontal bar plot")
plt.xlabel("computer languages")
plt.ylabel("marks")
plt.show()
```

```python
# histogram plot

import numpy as np

data=np.random.randn(1000)
```

```python
plt.hist(data,bins=20,color="green")
plt.title("hist plot")
plt.xlabel("frequency")
plt.ylabel("values")
plt.show()
```

```python
#pie plot
sizes = [40, 35, 25]
labels = ['Python', 'Java', 'C++']
```

```python
plt.pie(sizes,labels=labels,autopct='%1.1f%%', startangle=140)
plt.title("language usage")
plt.show()
```

```python
#box plot
data = [5, 8, 10, 15, 18, 20, 22, 25, 30]
```

```python
plt.boxplot(data)
plt.title("box plot example")
plt.ylabel("values")
plt.show()
```

```python
#subplot (allows multiple plots)
x = [1, 2, 3, 4]
y1 = [2, 4, 6, 8]
y2 = [1, 3, 5, 7]
```

```python
plt.subplot(1,2,1)
plt.plot(x,y1)
plt.title("plot-1")

plt.subplot(1,2,2)
plt.plot(x,y2)
plt.title("plot-2")

plt.show()
```

```python
#legend
x = [1, 2, 3, 4]
plt.plot(x, [1, 4, 9, 16], label='Square')
plt.plot(x, [1, 8, 27, 64], label='Cube')
plt.title("legend example")
plt.legend()
plt.show()
```

```python
#grid
x = [1, 2, 3, 4]
y = [2, 4, 6, 8]
```

```python
plt.plot(x,y)
plt.title("example for grid")
plt.grid()
plt.show()
```

```python
#plt.figure(custom size)
plt.figure(figsize=(6,4))
plt.plot([1, 2, 3], [2, 4, 6])
plt.title("Custom Figure Size")
plt.show()
```

```python
#saving graph
plt.plot([1, 2, 3], [2, 4, 6])
plt.title("Save Plot Example")
plt.savefig("plot_output.png")
plt.show()
```

```python
#error plot
x = [1, 2, 3, 4]
y = [2, 4, 6, 8]
error = [0.2, 0.4, 0.3, 0.5]

plt.errorbar(x, y, yerr=error, fmt='o', color='black', ecolor='red', capsize=6)
plt.title("Error Bar Example")
plt.show()
```

```python
#area between two curves
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.fill_between(x, y, color='skyblue', alpha=0.4)
plt.title("Fill Between Example")
plt.show()
```

```python
#annotations
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.annotate("Peak Point", xy=(5, 10), xytext=(3, 9),
            arrowprops={"facecolor": "black", "shrink": 0.05})
plt.show()
```
